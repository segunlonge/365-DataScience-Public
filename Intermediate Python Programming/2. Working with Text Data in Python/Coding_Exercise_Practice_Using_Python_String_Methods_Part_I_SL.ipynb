## Practice Using Python String Methods - Part I

```python
s = "Income Statement"

# 1. Use the .replace() method to substitute the word "Income" with "Cash Flow". 
# Store the output in a variable called s1.
s1 = s.replace("Income", "Cash Flow")

# 2. Display the content of s1.
print(s1)

# 3. Verify that s1 does not start with "Income".
print(s1.startswith("Income"))

# 4. Verify that s1 starts with "Cash Flow".
print(s1.startswith("Cash Flow"))

# 5. Verify that s1 does not end with "statement".
print(s1.endswith("statement"))

# 6. Verify that s1 ends with "Statement".
print(s1.endswith("Statement"))

# 7. Display the number of characters in s1.
print(len(s1))

# 8. Split the words in s1 using the .split() method with the appropriate argument. 
# Store the resulting list in list_01.
list_01 = s1.split(" ")

# 9. Display the content of list_01.
print(list_01)

# 10. Display number of elements in list_01.
print(len(list_01))
```
