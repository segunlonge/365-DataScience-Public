```python
import pandas as pd

car_info = ['Ford-USA (North America)',
              'Chevrolet-USA (North America)',
              'Cadillac-USA (North America)',
              'BMW-Germany (Europe)',
              'Peugeot-France (Europe)',
              'Fiat-Italy (Europe)',
              'Toyota-Japan (Asia)']
              
# 1. Convert the car_info list into a pandas Series, and reassign this new Series to car_info.
car_info = pd.Series(car_info)


# 2. Use the relevant string accessor twice—once with the .split() method and once to index the element in the newly obtained list—to isolate the car brand name (before the dash). 
# Assign this result to a new variable called car_brands. 
# Use the print() function to display the content of car_brands to verify it holds the expected data.
car_brands = car_info.str.split("-").str[0]
print(car_brands)

# 3. Using a similar approach, isolate the country and continent information (after the dash) and assign it to a new variable named origin. 
# Use the print() function to display the content of origin to verify it holds the expected data.
origin = car_info.str.split("-").str[1]
print(origin)

# 4. Use the print() function to check the data type of the origin variable.
print(type(origin))

# 5. Do all entries in origin relate to North America? To answer this question, 
# obtain a Series of Boolean values indicating True for brands from North America and False otherwise.
# Use the print() function to display the output.
print(origin.str.contains("North America"))
```
