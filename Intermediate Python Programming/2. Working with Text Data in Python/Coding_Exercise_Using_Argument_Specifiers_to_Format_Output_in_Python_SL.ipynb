## Using Argument Specifiers to Format Output in Python

```python
motorcycle_shop = ['New Motorcycles', 'Modern Motorcycles']
motorcycle_brand = ['Ducati', 'Aprilia', 'BMW', 'Suzuki', 'Honda', 'Yamaha', 'Kawasaki']
model_price_usd = [29000, 20000, 33000]

# 1. "Next month, the "New Motorcycles" shop is expected to sell the newest Ducati model at a price of $33000 per unit."
print("Next month, the \"%s\" shop is expected to sell the newest %s model at a price of $%d per unit." % 
(motorcycle_shop[0], motorcycle_brand[0], model_price_usd[2]))

# 2. By only changing the values associated to the specifiers, # modify the output as follows:
# - Change the shop name to "Modern Motorcycles"
# - Change the brand to BMW.
# - Keep the same price.
# Print the string again!
print("Next month, the \"%s\" shop is expected to sell the newest %s model at a price of $%d per unit." % 
(motorcycle_shop[1], motorcycle_brand[2], model_price_usd[2]))

#3. By only changing the values associated to the specifiers, modify the output as follows:
# - Change the brand to Aprilia
# - Change the price to $29000
# - Keep the same shop.
# Print the string again!
print("Next month, the \"%s\" shop is expected to sell the newest %s model at a price of $%d per unit." % 
(motorcycle_shop[1], motorcycle_brand[1], model_price_usd[0]))
```
