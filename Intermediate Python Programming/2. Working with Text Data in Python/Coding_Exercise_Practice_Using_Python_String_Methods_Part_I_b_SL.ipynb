```python
financial_statement = "   Income statement "

# 1. Use a Python method to convert all the letters in the string to lowercase and display the result.
print(financial_statement.lower())

# 2. Use a Python method to convert all the letters in the string to uppercase and display the result.
print(financial_statement.upper())

# 3. Apply a Python method that will return the first letter of each word in financial_statement as uppercase, 
# while the rest of the letters remain lowercase.
print(financial_statement.title())

# 4. Use a Python string method to remove the whitespaces on the left side of the string. Display the result.
print(financial_statement.lstrip())

# 5. Use a Python string method to remove the whitespaces on the right side of the string. Display the result.
print(financial_statement.rstrip())

# 6. Return a copy of financial_statement with both leading and trailing whitespace characters removed.
print(financial_statement.strip())

# 7. Chain two Python methods to remove all leading and trailing whitespaces of financial_statement, 
# and then convert the first letter of each word to uppercase while maintaining the rest of the letters lowercase.
print((financial_statement.strip()).title())
```
