```python
event_cost = [100, 120, 140]
book_unit_price_usd = [35, 40, 42]

for event_cost_per in event_cost:
    for unit_price in book_unit_price_usd:
        for book in range(1, 10):
            print('With total costs of ${total_cost}, the expected profit from selling {book} book(s) at a price of ${unit_price} is: ${profit}'.format
                  (total_cost = event_cost_per,
                   book=book,
                   unit_price = unit_price,
                   profit = book * unit_price - event_cost_per))
            
```
