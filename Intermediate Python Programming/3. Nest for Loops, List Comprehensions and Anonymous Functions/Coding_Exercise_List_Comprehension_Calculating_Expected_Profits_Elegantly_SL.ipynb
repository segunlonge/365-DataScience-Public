```python
event_cost = [100, 120, 140]
book_unit_price_usd = [35, 40, 42]

[
    print("With total costs of ${}, the expected profit from selling {} book(s) at a price of ${} is: ${}".format(
        total_cost, books_sold, unit_price, (unit_price * books_sold) - total_cost))
    for total_cost in event_cost
    for books_sold in range(1, 10)
    for unit_price in book_unit_price_usd
]
```
