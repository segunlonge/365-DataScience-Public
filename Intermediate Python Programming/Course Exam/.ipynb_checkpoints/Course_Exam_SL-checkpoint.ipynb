#### Question 4

```python
departments = ['dept-a1-04', 'dept-a2-54', 'dept-b1-79', 'dept-c-22']
exp_net_sales = [1300, 1500, 23080, 1000, 11050]
exp_cost_of_sales = [3000, 3500]

for dept in departments:
    for exp_ns in exp_net_sales:
        for exp_cos in exp_cost_of_sales:
            print('Next year\'s expected profit for department {0} is ${1} mln.'.format(dept, exp_ns - exp_cos))
            

```
