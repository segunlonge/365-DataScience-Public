```python
# 0. Import the relevant libraries
import numpy as np

# 1. Create the first list
list_1 = [2,4,6]
print(list_1)

# 2. Create the second list
list_2 = [20,40,60]
print(list_2)

# 3. Create the matrix
matrix = np.array([list_1,list_2])
print(matrix)

# 4. Check the type and shape of the matrix
print(type(matrix)) # Type of the matrix variable
print(matrix.shape) # Shape of the matrix variable

# 5. Combine the two lists
list_sum = list_1 + list_2
print(list_sum)

# 6. Find the square root of the matrix values
matrix_sqrt = np.sqrt(matrix)
print(matrix_sqrt)
```
