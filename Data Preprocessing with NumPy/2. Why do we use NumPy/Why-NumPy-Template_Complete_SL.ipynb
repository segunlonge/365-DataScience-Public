### List vs Array

```python
list_a = [1,2,3,4,5,6] # list
```

```python
len(list_a)
```

```python
type(list_a)
```

```python
list_b = [[1,2,3],[4,5,6]] # list of lists
```

```python
len(list_b)
```

```python
array_b = np.array(list_b)
```

```python
type(array_b)
```

```python
print(list_b)
```

```python
print(array_b)
```

```python
array_b.shape
```

```python
len(list_b) # shape function cannot be used on a list. Instead use len to get the 'shape' e.g. this list is made of 2 elements
```

```python
len(list_b[0]) # this is the number of elements in the first list
```

```python
list_c = list_b[0] + list_b[1] # this operation concatenates the first list with the second list
array_c = array_b[0] + array_b[1] # adds corresponding elements from the 1st and 2nd rows naturally
```

```python
print(list_c)
print(array_c)
```

```python
# performing math functions on array can be done with another library such as math but can only be done on an element by element basis
import math
sqroot = math.sqrt(array_b[1,0])
print(sqroot)
```

```python
# or alternatvely use the native math functions in numpy and the math operation will be performed on each element in the array
# lists don't have this element wise functionality
np.sqrt(array_b)
```
