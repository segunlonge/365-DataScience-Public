```python
import numpy as np

# Since each face as an equal chance of landing up. P=1/20. n = 100 trials
np.random.seed(42)
n = 100  # total rolls
p = 1/20  # probability of rolling a '13'
count_13 = np.random.binomial(n, p)
print(count_13)

# alternative check by generating 100 trials of throwing the die
rolls = np.random.randint(1, 21, n)
print(rolls)
count_13_b = np.sum(rolls == 13)
print(count_13_b)
```
