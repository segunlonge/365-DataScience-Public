# Generating Data w/ Numpy

```python
import numpy as np
```

### np.empty(), np.zeros(), np.ones(), np.full()

```python
# creates an empty array of a defined shape with empty values. This is the fastest way to create an nd array
array_empty = np.empty(shape = (2,3))
array_empty
```

```python
# creates an empty array of a defined shape with empty values. The advantage over np.empty() is that values are always 
# 0 rather than some random tiny numbers

array_0s = np.zeros(shape = (2,3))
array_0s
```

```python
# creating an empty nd array using np.zeros but with a defined datatype

array_0s = np.zeros(shape = (2,3), dtype = np.int8)
array_0s
```

```python
# works just like np.zeros but with 1ss instead of zeros

array_1s = np.ones(shape = (2,3))
array_1s
```

```python
# works like the other empty array values except that the fill_value parameter is required
# fill_value values can be other data types like string too

array_full = np.full(shape = (2,3), fill_value = 2)
array_full
```

### "_like" functions

```python
matrix_A = np.array([[1,0,9,2,2],[3,23,4,5,1],[0,2,3,4,1]])
matrix_A
```

```python
# like functions are equvilent to the previous functions where the definition of the shape is based on another existing matrix
array_empty_like = np.empty_like(matrix_A)
array_empty_like
```

```python
# here we copy the shape of another matrix but fill it with 0s instead of the random values in the empty_like function
array_0s_like = np.zeros_like(matrix_A)
array_0s_like

# applications:
# can be used as a form of switch based on another array of the same shape
```

### np.arange()

```python
# generate a sequence of numbers within a range as a list
list(range(30))
```

```python
# generate a sequence of numbers within a range as an array
array_rng = np.arange(30)
array_rng
```

```python
# generate a sequence of numbers within a range as an array (with useful arguments)
array_rng = np.arange(stop = 30)
array_rng
```

```python
# generate a sequence of numbers within a range as an array (with useful arguments)
array_rng = np.arange(start = 0, stop = 30)
array_rng
```

```python
# generate a sequence of numbers within a range as an array (with useful arguments)
array_rng = np.arange(start = 0, stop = 30, step = 2.5)
array_rng
```

```python
# generate a sequence of numbers within a range as an array (with useful arguments)
array_rng = np.arange(start = 0, stop = 30, step = 2.5, dtype = np.float64)
array_rng
```

## Random Generators

### Defining Random Generators

```python
from numpy.random import Generator as gen # the Generator function takes a bit generator as an input and creates generator objects
from numpy.random import PCG64 as pcg # Permutation Congruential Generator
```

```python
array_RG = gen(pcg())
array_RG.normal() # random numbers from a normal distribution
```

```python
# the size argument allows us to specify how many random values we wish to generate
array_RG = gen(pcg())
array_RG.normal(size = 5)
# this is a 1-d nd array
```

```python
# the size argument allows us to specify how many random values we wish to generate
array_RG = gen(pcg())
array_RG.normal(size = (5,5))
# this is a 5x5 nd array specified inside the tuple
```

### Generating Integers, Probabilities and Random Choices

```python
# the size argument allows us to specify how many random values we wish to generate
# to keep consistant random numbers specify a seed value which would give the same random numbers on every generation
array_RG = gen(pcg(seed = 365))
array_RG.normal(size = (5,5))
```

```python
# generate random integers between 0 and 10 of 5x5 nd arrays
array_RG = gen(pcg(seed = 365))
array_RG.integers(10, size = (5,5))
```

```python
#...or specify upper and lower limits
array_RG = gen(pcg(seed = 365))
array_RG.integers(low = 10, high = 100, size = (5,5))
```

```python
# generating random probability values (from a continous uniform distribution)
array_RG = gen(pcg(seed = 365))
array_RG.random(size = (5,5))
```

```python
# generating random probability values (from a range of numbers from ones choice)
# this method defaults to a unform distribution i.e. each element has an equal probability of being selected
array_RG = gen(pcg(seed = 365))
array_RG.choice([1,2,3,4,5], size = (5,5))
```

```python
# generating random probability values (from a range of numbers from ones choice)
# this method defaults to a unform distribution i.e. each element has an equal probability of being selected
# however, using the p argument, the probability for each element can be inputted
array_RG = gen(pcg(seed = 365))
array_RG.choice([1,2,3,4,5], p = [0.1,0.1,0.1,0.1,0.6], size = (5,5))
```
