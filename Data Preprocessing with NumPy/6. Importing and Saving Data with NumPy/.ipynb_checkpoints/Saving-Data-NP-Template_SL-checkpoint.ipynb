## Saving Files with NumPy

```python
import numpy as np
```

### np.save()

```python
lending_co = np.genfromtxt("Lending-Company-Saving.csv",
                           delimiter = ',',
                           dtype = str)

print(lending_co)
```

```python
# the np.save() method creates an npy file in the same directory (folder) as your notebook document
# npy is a special type of text file native to jupyter notes (much faster than and take up less memory than a regualr csv)
np.save("Lending-Company-Saving", lending_co)
```

```python
# loading back the npy file and assigning it to a variable
lending_data_save = np.load("Lending-Company-Saving.npy")
```

```python
# Printing the npy file (compare it to the csv version)
print(lending_data_save)
```

```python
# do an objective comparison (csv vs npy file) with np.array_equal which returns True as they are both structurally the same
np.array_equal(lending_data_save, lending_co)
```

### np.savez()

```python
# load the lending company saving file from csv and save as npy
lending_co = np.genfromtxt("Lending-Company-Saving.csv",
                           delimiter = ',',
                           dtype = str)

lending_data_save = np.load("Lending-Company-Saving.npy") # from previous exercise
```

```python
# np.savez creates an npz file instead of an npy
# it's like an archive of npys that can store multiple arrays
# instead of storing different datasets in seperate npy files, we can
# store all of them in a single npz
np.savez("Lending-Company-Saving", lending_co, lending_data_save) # here we are loading two files (one csv and one npy) into an array
```

```python
lending_data_savez = np.load("Lending-Company-Saving.npz")
```

```python
print(lending_data_savez["arr_0"])

# however, note that the whole npz file cannot be printed as it consists of multiple datasets
# so the array index of the dataset you want to print has to be specified 
```

```python
# rather than stick to the default array names assgined by numpy distinct array names can be assigned to each dataset
np.savez("Lending-Company-Saving", company = lending_co, data_save = lending_data_save)
```

```python
# re-load again
lending_data_savez = np.load("Lending-Company-Saving.npz")
```

```python
# check the custom arrays defined above
lending_data_savez.files
```

```python
# printing the first array using the user defined array name rather than the numpy assigned name
print(lending_data_savez["company"])
```

```python
# comparing the csv and npz version of the same file reveals that they are both the same
np.array_equal(lending_data_savez["company"],lending_data_savez["data_save"])
```

### np.savetxt()

```python
# load the dataset for use with the np.savetxt function in the next line
lending_co = np.genfromtxt("Lending-Company-Saving.csv",
                           delimiter = ',',
                           dtype = str)
```

```python
# npsavetxt() function helps store NumPy datasets in text files in either txt or csv
np.savetxt("Lending-Company-Saving.txt", 
           lending_co,
            fmt = '%s', 
            delimiter = ',')
```

```python
# Re-load the data that was saved with the npsavetxt function
lending_data_savetxt = np.genfromtxt("Lending-Company-Saving.txt",
                           delimiter = ',',
                           dtype = str)

print(lending_data_savetxt)
```

```python
# load the previous npy version of the file
lending_data_save = np.load("Lending-Company-Saving.npy")
```

```python
# compare the text version using the npsavetxt function with the npy version created at the beginning 
np.array_equal(lending_data_savetxt, lending_data_save)
```
