## Importing Data with NumPy

```python
import numpy as np
```

### np.loadtxt() vs np.genfromtxt()

```python
lending_co_data_numeric_1 = np.loadtxt("Lending-Company-Numeric-Data.csv", delimiter = ',')
lending_co_data_numeric_1
```

```python
lending_co_data_numeric_2 = np.genfromtxt("Lending-Company-Numeric-Data.csv", delimiter = ',')
lending_co_data_numeric_2
```

```python
# proof that loadtxt and genfromtxt functions generated identical datasets
np.array_equal(lending_co_data_numeric_1, lending_co_data_numeric_2)

# loadtxt is faster but cannot handle missing values
```

```python
lending_co_data_numeric_NAN = np.loadtxt("Lending-Company-Numeric-Data-NAN.csv", delimiter = ';')
lending_co_data_numeric_NAN

# This causes an error due to missing values in certain columns
```

```python
lending_co_data_numeric_NAN_2 = np.genfromtxt("Lending-Company-Numeric-Data-NAN.csv", delimiter = ';')
lending_co_data_numeric_NAN_2

# genfromtxt is able to handle missing data
```

```python
lending_co_data_numeric_NAN_3 = np.genfromtxt("Lending-Company-Numeric-Data-NAN.csv", 
                                              delimiter = ';',
                                             dtype = str)
lending_co_data_numeric_NAN_3

# a way of getting around missing values using the genfromtxt function is to set the datatype to string.
# however doing this means that no mathematical operations can then be performed on the imported data set.
```

### Partial Cleaning While Importing

```python
lending_co_data_numeric_NAN_4 = np.genfromtxt("Lending-Company-Numeric-Data-NAN.csv", delimiter = ';')
lending_co_data_numeric_NAN_4
```

```python
# skip first two rows etc...
lending_co_data_numeric_NAN_5 = np.genfromtxt("Lending-Company-Numeric-Data-NAN.csv", delimiter = ';',
                                                 skip_header = 2)
lending_co_data_numeric_NAN_5
```

```python
# skip last two rows etc...
lending_co_data_numeric_NAN_6 = np.genfromtxt("Lending-Company-Numeric-Data-NAN.csv", delimiter = ';',
                                                 skip_footer = 2)
lending_co_data_numeric_NAN_6
```

```python
# display only selected columns by column number etc...and the order of the columns canbe changed by re-arranging the column numbers
lending_co_data_numeric_NAN_7 = np.genfromtxt("Lending-Company-Numeric-Data-NAN.csv", delimiter = ';',
                                                 usecols = (0,1,5))
lending_co_data_numeric_NAN_7
```

```python
# usecols and skip_header arguments can be combined....
lending_co_data_numeric_NAN_8 = np.genfromtxt("Lending-Company-Numeric-Data-NAN.csv", delimiter = ';',
                                                 usecols = (0,1,5),
                                                 skip_header = 2,
                                                 skip_footer = 2)
lending_co_data_numeric_NAN_8
```

```python
# elements of an array can be assigned to separate variables....
lending_co_data_0, lending_co_data_1, lending_co_data_5 = np.genfromtxt("Lending-Company-Numeric-Data-NAN.csv", delimiter = ';',
                                                 usecols = (0,1,5),
                                                 skip_header = 2,
                                                 skip_footer = 2,
                                                 unpack = True)
print(lending_co_data_0)
print(lending_co_data_1)
print(lending_co_data_5)
```

### String vs Object vs Numbers

```python
# importing raw
lending_co_lt = np.genfromtxt("lending-co-LT.csv", 
                                delimiter = ',')
print(lending_co_lt)
```

```python
# importing as integers
lending_co_lt = np.genfromtxt("lending-co-LT.csv", 
                                delimiter = ',',
                                 dtype = np.int32)
print(lending_co_lt)
```

```python
# importing as strings
lending_co_lt = np.genfromtxt("lending-co-LT.csv", 
                                delimiter = ',',
                                 dtype = str)
print(lending_co_lt)
```

```python
# importing as objects; for better compatibility with the older versions of the pandas package
lending_co_lt = np.genfromtxt("lending-co-LT.csv", 
                                delimiter = ',',
                                 dtype = object)
print(lending_co_lt)
```
