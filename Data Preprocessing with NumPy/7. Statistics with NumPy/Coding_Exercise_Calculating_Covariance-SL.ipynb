```python
import numpy as np

sample_array = np.array([[-6, -1, -1, 7], [3, 4, 2, 2], [5, 2, 3, 4], [4, -3, 5, -1]])

# Find the covariance between all rows with np.cov()
# The covariance between the 2nd and 4th row 
# is the element on the 2nd row and 4th column of np.cov()
print(np.cov(sample_array)[1,3]) # (Symmetrically, the 4th row and 2nd column also work)
```
