```python
import numpy as np

array_A = np.array([[-2, 1, 0], [6, -3, 0], [-4, -6, -1]])
array_B = np.array([[-3, 2, 1], [1, -1, 0], [-3, 4, 5]])

# element-wise maximum of array_A and array_B
combined_array = np.concatenate([[np.max(array_A, axis = 0)], [np.max(array_B, axis = 0)]])
print(combined_array)
```

```python
# peak-to-peak differences for each column in combined_array
print(np.ptp(combined_array, axis = 0))
```

```python
# the nearest element to the 74th percentile
print(int(np.percentile(combined_array, 74, method = "nearest")))
```

### Solution (Correct)

```python
import numpy as np

array_A = np.array([[-2, 1, 0], [6, -3, 0], [-4, -6, -1]])
array_B = np.array([[-3, 2, 1], [1, -1, 0], [-3, 4, 5]])

# element-wise maximum of array_A and array_B
combined_array = np.maximum(array_A, array_B)
print(combined_array)

# peak-to-peak differences for each column in combined_array
print(np.ptp(combined_array, axis = 0))

# the nearest element to the 74th percentile
print(int(np.percentile(combined_array, 74, method = "nearest")))
```
