# Statistsics w/ NumPy

```python
import numpy as np
```

### np.mean()

```python
matrix_A = np.array([[1,0,0,3,1],[3,6,6,2,9],[4,5,3,8,0]])
matrix_A
```

```python
# note that to calculate the mean from the 3 by 5 array the matrix is flattened into a 1-d array
float(np.mean(matrix_A)) # note: wrapping the function around float returns a normal number instead of np.float(3.4)
```

```python
# to find the mean of the first row use array index
float(np.mean(matrix_A[0])) 
```

```python
# to find the mean of the first column use array index
float(np.mean(matrix_A[:,0])) 
```

```python
# to find the mean of the every column without writing the code for each column index
np.mean(matrix_A, axis = 0) # the axis = 0 argument allows the mean to be calculated over the columns
```

```python
# to find the mean of the every row without writing the code for each column index
np.mean(matrix_A, axis = 1) # the axis = 1 argument allows the mean to be calculated over the rows
```

```python
# instead of calling the function to calculate the mean the method can also be used
float(matrix_A.mean())

# however it is preferable to use the function rather than the method because there are more functions of the equivalent methods
# so using functions can help keep a project looking consistent
# functions also offer more control over the equivalent method (as shown in the next cell below)
```

```python
# in this case using the mean function rather than the method offers broadcastability, axis option and option to set the data type
np.mean(matrix_A, axis = 1, dtype = np.int64)
```

### Min & Max Values

```python
import numpy as np
```

```python
matrix_A = np.array([[1,0,0,3,1],[3,6,6,2,9],[4,5,3,8,0]])
matrix_A
```

```python
# the min function finds the lowest value along the given axis else if there is no axis specified it flattens the array and gives the lowest value
float(np.min(matrix_A))
```

```python
# with second row specified 
float(np.min(matrix_A[1]))
```

```python
# using the amin function just means finding the lowest value specifically for an array
# min() and amin() are identical functions
float(np.amin(matrix_A))
```

```python
# the minimum function holds element-wise minium values
# this example produces the mimimum values from the first and third row
np.minimum(matrix_A[0], matrix_A[2])
```

```python
# comparing more than 2 rows? use the reduce property
# this example assigns each row to an array from comparison
np.minimum.reduce(matrix_A)
```

```python
# the min function over the zero axis also produces the same output as minimum.reduce both methods are equivalent 
np.min(matrix_A, axis = 0)
```

```python
# the max equivalents of the above
float(np.max(matrix_A))
```

```python
# the max equivalents of the above
float(np.amax(matrix_A))
```

```python
# the max equivalents of the above
np.maximum.reduce(matrix_A)
```

### Statistical Order Functions 

```python
import numpy as np
```

```python
matrix_A = np.array([[1,0,0,3,1],[3,6,6,2,9],[4,5,3,8,0]])
matrix_A
```

```python
# returns the difference between the highest and lowest value in an array
# used instead of combining the min and max functions
float(np.ptp(matrix_A))
```

```python
# for each element in the row the difference between the highest and lowest
np.ptp(matrix_A, axis = 0)
```

```python
# for each element in the column the difference between the highest and lowest
np.ptp(matrix_A, axis = 1)
```

```python
# sorts all the elements of the array matrix_A into ascending order and returns a flattened 1D array
np.sort(matrix_A, axis = None)
```

```python
# return the value of the 70th percentile
# a percentile is a value that is greater than the corresponding % of the dataset
float(np.percentile(matrix_A, 70))

# in this example the 70th percentile is an intermediate value that lies somewhere between 4 and 5
```

```python
# a more precise value can be attaind by setting the returned value as a midpoint
# if the percentile is between two values
float(np.percentile(matrix_A, 70, method = "midpoint"))
```

```python
# ...alternatively the higher of the two values can be returned
float(np.percentile(matrix_A, 70, method = "higher"))
```

```python
# ...alternatively the lower of the two values can be returned
float(np.percentile(matrix_A, 70, method = "lower"))
```

```python
# ...alternatively the nearest rounds the intermmediate value to the nearest of the two
float(np.percentile(matrix_A, 70, method = "nearest"))
```

```python
# The percentile function can be used to calculate the median value of the data set
float(np.percentile(matrix_A, 50))
```

```python
# ... or it can be used to calculate the minimum value of the data set
float(np.percentile(matrix_A, 0))
```

```python
# ... or the maximum value of the data set
float(np.percentile(matrix_A, 100))
```

```python
# quantile and percentile are the same except that quantile is between 0 and 1 and percentile is between 0 and 100
float(np.quantile(matrix_A, 0.7))
```

```python
# the arguments of quantile and percentile function are the same
float(np.quantile(matrix_A, 0.7, method = "midpoint"))
```

### Averages and Variances

```python
import numpy as np
```

```python
matrix_A = np.array([[1,0,0,3,1],[3,6,6,2,9],[4,5,3,8,0]])
matrix_A
```

```python
# median is the middle value of a sorted dataset
float(np.median(matrix_A))
```

```python
# the sort function confirms the median value
np.sort(matrix_A, axis = None)
```

```python
# the mean function calculates the arithmetic average of all the values in the dataset
float(np.mean(matrix_A))
```

```python
# works like the mean function but there are sublte differences (it can calculated averages with weights)
float(np.average(matrix_A))
```

```python
# This code snippet uses NumPy’s modern random number generation API to produce a 3 x 5 matrix of pseudorandom numbers 
# uniformly distributed in the half-open interval $[0.0, 1.0)$.

from numpy.random import Generator as gen # transforms the raw stream of random bits from PCG64 into numbers from specific statistical distributions
from numpy.random import PCG64 as pcg # produces a raw stream of random bits

array_RG = gen(pcg(365))

array_weights = array_RG.random(size = (3,5)) # continous random floating point values betweem [0.1,1.0)
array_weights
```

```python
# claculates the average of matrix A with the weights
float(np.average(matrix_A, weights = array_weights))
```

```python
# calculates the varience of matrix A
float(np.var(matrix_A))
```

```python
# calculates the standard deviation of matrix A
float(np.std(matrix_A))
```

### Correlation

```python
import numpy as np
```

```python
matrix_A = np.array([[1,0,0,3,1],[3,6,6,2,9],[4,5,3,8,0]])
matrix_A
```

```python
# covariance is a statistical measure that tells us how two variables change together
# the covariances between different rows of matrix A
np.cov(matrix_A)

# the main diagonal elements of the covariance matrix represent the variance of each row, as the covariance between an element 
# and itself is simply its variance.
```

```python
# correlation between every different rows in the array so there should be nine correlation coefficients
np.corrcoef(matrix_A)
```

### Histograms

```python
import numpy as np
```

```python
matrix_A = np.array([[1,0,0,3,1],[3,6,6,2,9],[4,5,3,8,0]])
matrix_A
```

```python
# a histogram is a way to examine a dataset by dissecting its density i.e. how many values from a dataset fall within a 
# pre-determined range i.e. called bins
```

```python
# flattening and sorting the arrays will help with visualising how the bins work
np.sort(matrix_A, axis = None)
```

```python
# the histogram function sorts the values in default bins
# first array is the count of the values per bin
# second array are the bin edges
# the bins are closed-open intervals i.e. [a,b) a will fall between a and b and b would not
np.histogram(matrix_A)
```

```python
# since the output of the function is a tuple; we can use indexing to extract each element
# e.g. to extract the density add the index
np.histogram(matrix_A)[0]
```

```python
# since the output of the function is a tuple; we can use indexing to extract each element
# e.g. to extract the bin edges add the index
np.histogram(matrix_A)[1]
```

```python
# the default bin size is 10 but this can be changed

np.histogram(matrix_A, bins = 4)
```

```python
# the range argument sets the range of values placed in the bins
np.histogram(matrix_A, bins = 4, range = (1,7))
```

```python
# displaying the results of the histogram graphically
import matplotlib.pyplot as plt
plt.hist(matrix_A.flat, bins = np.histogram(matrix_A)[1])
plt.show()
```

```python
matrix_A
```

```python
# for the purposes of this exercise we want to plot a 2-d histogram by getting the x and y coordiinates as row
# 0 and 1 of Matrix A

np.histogram2d(matrix_A[0], matrix_A[1], bins = 4)

# this produces the 2d array of the density of the values and the bins for the x and y cordinate
# illustration: the 2 in the first row of the array is number in the first block on the x axis from the x (the two 0s) row in the matrix
# and the y row in the matrix (the two 6s)
```

```python
# this is the histogram equivalent of the 2d one above but for higher dimensions
np.histogramdd(matrix_A.transpose(), bins = 4)

# this means that there is a 2d density array for each bin of the z-coordinate array
```

### NaN-Equivalents

```python
import numpy as np
```

```python
matrix_A = np.array([[1,0,0,3,1],[3,6,6,2,9],[4,5,3,8,0]])
matrix_A
```

```python
# NAN-equivalent functions are functions that work on arrays with missing values in them
float(np.nanmean(matrix_A))
```

```python
# The non NAN-equivalent function of mean returns the same result as the NAN-equivalent
float(np.mean(matrix_A))

```

```python
# np.nan means elements that are not a number
matrix_B = np.array([[1,0,0,3,1],[3,6,np.nan,2,9],[4,5,3,8,0]])
matrix_B
```

```python
# the nan.mean function can handle nan values
float(np.nanmean(matrix_B))
```

```python
# the ordinary mean function cannot handle nan values
float(np.mean(matrix_B))
```

```python
# these NAN-equivalent functions can also handle nan values
float(np.nanquantile(matrix_B, 0.7))
```

```python
# these NAN-equivalent functions can also handle nan values
float(np.nanvar(matrix_B))
```
