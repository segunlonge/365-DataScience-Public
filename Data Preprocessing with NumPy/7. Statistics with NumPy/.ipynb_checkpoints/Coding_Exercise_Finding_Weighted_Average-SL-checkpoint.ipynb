```python
import numpy as np

sample_array = np.array([3, -1, 13, 10, 4, 7, 0, -2, -11, 16])
sample_weights = np.array([1, 1.5, 2, 0.5, 1.2, 3, 1, 4, 2, 2.6])

median_array = np.median(sample_array) 
average_array = np.average(sample_array, weights = sample_weights)

print(median_array)
print(average_array)

```
