```python
import numpy as np

# Array containing the [x, y, z] coordinates of 6 objects
coordinates = np.array([[3, 0.4, -1], [1.8, -2.3, -0.9], [6, 4, 8], [-1, 2.6, 4.2], [0, 0, 12], [-5.3, -2, -6.9]])

# Print the rows for which y < 2
print("Task 1:\n", coordinates[coordinates[:,1] < 2]) 

# Print the rows for which all coordinates are positive
# You can store the condition in a different variable 'mask' to make it more readable
mask = (coordinates > 0).all(axis=1) # axis=1 means work across the columns of each row.
print("\nTask 2:\n", coordinates[mask])
```
