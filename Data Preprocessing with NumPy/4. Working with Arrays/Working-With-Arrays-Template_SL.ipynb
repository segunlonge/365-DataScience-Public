# Working with Arrays

```python
import numpy as np
```

## Slicing

```python
matrix_A = np.array([[1,2,3],[4,5,6]])
matrix_A
```

### Basic Slicing

```python
matrix_A[:] # default slicing returns the exact same matrix back because no before or after rows have been specified
```

```python
matrix_A[0:0] # returns an empty matrix because there are no rows the first one and itself
```

```python
matrix_A[0:1] # returns the rows between the first row and the second row i.e. the first row

# note: indexing in numpy is closed-open meaning the upper limit is not included when slicing
```

```python
matrix_A[1:] # take from the second row onwards
```

```python
matrix_A[:2] # take upto the second row
```

```python
matrix_A[0] # produces a 1-d slice
```

```python
matrix_A[:1] # same putput but produces a 2-d slice
```

```python
matrix_A[:-1] # the use of negative values e.g. taking the element prior to the last value
```

```python
matrix_A[:,1:] # slicing columns, to the left of the comma are the rows and to the right of the comma are the columns

# this example takes all rows but then slices from the second column onwards
```

```python
matrix_A[1:,1:] # slice from second column and second row onwards
```

### Stepwise Slicing

```python
matrix_B = np.array([[1,1,1,2,0],[3,6,6,7,4],[4,5,3,8,0]])
matrix_B
```

```python
# slicing works by: first colon separates the row and column and the second colon separates the step.
# e.g. the below matrix takes all rows and columns but skips every other row
matrix_B[::2,::]
```

```python
# slicing works by: first colon separates the row and column and the second colon separates the step.
# e.g. the below matrix takes all rows and columns but skips every other column
matrix_B[::,::2]
```

```python
# slicing works by: first colon separates the row and column and the second colon separates the step.
# e.g. the below matrix takes all rows and columns but skips every other row and every other column
matrix_B[::2,::2]
```

```python
# dealing with negative steps; skips rows and/or columns but works backwards as can be seen below
# slicing works by: first colon separates the row and column and the second colon separates the step.
# e.g. the below matrix takes all rows and columns but skips every other row and every other column
matrix_B[::-2,::2]
```

### Conditional Slicing

```python
matrix_c = np.array([[1,1,1,2,0],[3,6,6,7,4],[4,5,3,8,0]])
matrix_c
```

```python
matrix_c[:,0] # slice only all the rows but only the first column
```

```python
matrix_c[:,0] > 2 # applying conditions to the array elements
```

```python
matrix_c[:,:] > 2 # apply the condition to the entire array i.e. all rows and all columns
```

```python
matrix_c[matrix_c[:,:] > 2] # displaying the values that meet the conditions instead of True or False
# since NumPy doesn't know how many elements would fit the condition it outputs a 1-D array instead
```

```python
# multiple conditions can also be applied to filtering an n-d array
matrix_c[(matrix_c[:,:] % 2 == 0) & (matrix_c[:,:] <= 4)]
# this condition filters the matrix for even numbers AND which are equal to or less than 4
```

```python
# multiple conditions can also be applied to filtering an n-d array
matrix_c[(matrix_c[:,:] % 2 == 0) | (matrix_c[:,:] <= 4)]
# this condition filters the matrix for even numbers OR that are equal to or less than 4
```

## Dimensions and the Squeeze Function

```python
matrix_D = np.array([[1,1,1,2,0],[3,6,6,7,4],[4,5,3,8,0]])
matrix_D
```

```python
# the different ways of indexing matrix D give different output shapes: scalar; vector; matrix
print(matrix_D[0,0].shape)
print(matrix_D[0,0:1].shape)
print(matrix_D[0:1,0:1].shape)

# the squeeze method removes all the unecessary dimensions of an array
```

```python
# squeeze method returns a single numeric value which is a 0-D array rather than a scalar
matrix_D[0:1,0:1].squeeze()
```

```python
# this removes the unescessary dimesions associated with the matrix shape even if only returning one value
matrix_D[0:1,0:1]
```

```python
# the squeezed array of previously 1 by 1 dimension now has no shape
matrix_D[0:1,0:1].squeeze().shape
```

```python
# an equivelent way of using the squeeze functions
np.squeeze(matrix_D[0:1,0:1]).shape
```

```python
# By using the squeeze method all the outputs of the different matrix are converted into equivelent outputs
print(matrix_D[0,0].squeeze().shape)
print(matrix_D[0,0:1].squeeze().shape)
print(matrix_D[0:1,0:1].squeeze().shape)
```
