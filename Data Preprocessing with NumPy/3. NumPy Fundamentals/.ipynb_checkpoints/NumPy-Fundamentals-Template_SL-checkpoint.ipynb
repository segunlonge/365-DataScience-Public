# NumPy Fundamentals

```python
import numpy as np
```

## Indexing

```python
array_a = np.array([[1,2,3],[4,5,6]])
array_a
```

### Specific Values

```python
array_a[0] # first element of the 2d array
```

```python
array_a[1] # second element of the 2d array
```

```python
array_a[0][1] # first row and second element of first row
```

```python
array_a[0,1] # an alternative war to get the first row and second element of first row
```

```python
array_a[:,0] # slicing i.e. get all the rows of the first column
```

### Negative Indices

```python
array_b = np.array([1,2,3])
array_b[-1] # get the last element of the array. It's like going backwards from 0 which is the first element to -1 which is the last element
```

```python
array_a[-1] # for a multi-dimensional array as in this example -1 represents the last element which is the last row
```

```python
array_a[-1,-1] # get the last row and last column
```

## Assigning Values

```python
array_a = np.array([[1,2,3],[4,5,6]])
array_a
```

```python
# changing an element of an array
array_a[0,2] = 9
array_a
```

```python
# changing an element of the array (but this time the entire first row)
array_a[0] = 9
array_a
```

```python
# changing an element of the array (but this time the entire first column)
array_a[:,0] = 9
array_a
```

```python
# changing elements of an array with list values
list_a = [8,7,8]

array_a[0] = list_a # referring to the first row of the matrix array
array_a
```

```python
# change all the elements in the array/matrix to one value
array_a[:] = 9
array_a
```

```python
# CAUTION: setting the matrix array to a single value overwrites the matrix structure
array_a = 9
array_a
```

## Elementwise Properties

```python
array_a = np.array([7,8,9])
array_a
```

```python
array_b = np.array([[1,2,3],[4,5,6]])
array_b
```

```python
array_a + 2 # element wise array. Adding 2 to each element in array_a
```

```python
array_b + 2 # element wise array (works for higher dimensional arrays). Adding 2 to each element in array_b
# element wise array works for all the other arithmetic operations
```

```python
# element wise operations do not work for list because the purpose of list is to store data whereas the pupose of arrays is to perform calculations
# Example

list_a = [1,2,3]
list_a + [2]
```

```python
# adding elements from two different shaped arrays (specify the elements to make them compatible with the same shape)
array_a + array_b[0]
```

```python
# the shapes of the arrays need to be compatible to perform element wise e.g. the length of the 1d array should be the same 
# as the length of the 2d array

array_a * array_b

# each element in array_a multiples each element in array_b on a row by row basis
```

## Types of Data Supported by NumPy

```python
array_a = np.array([[1,2,3],[4,5,6]])
array_a # default data types of arrays are set to None
```

```python
# changing data type of nd array from the default None
array_a = np.array([[1,2,3],[4,5,6]], dtype = "float32")
array_a 
```

```python
# alternative way of changing data type of nd array from the default None
# it is good programming practice to stay consistant with ont way
array_a = np.array([[1,2,3],[4,5,6]], np.float32)
array_a 
```

```python
# alternative way of changing data type of nd array from the default None
array_a = np.array([[1,2,3],[4,5,6]], np.complex64) # complex data type
array_a 
```

```python
# alternative way of changing data type of nd array from the default None
array_a = np.array([[1,0,3],[4,5,6]], np.bool) # bollean data type, anything but 0 is true
array_a
```

```python
# alternative way of changing data type of nd array from the default None
array_a = np.array([[1,0,3],[4,5,6]], str) # string data type (elements are enclosed in single quotes)
array_a
```

### Broadcasting

```python
# vector
array_a = np.array([1,2,3])
array_a
```

```python
# vector
array_b = np.array([[1],[2]])
array_b
```

```python
# matrix
matrix_c = np.array([[1,2,3],[4,5,6]])
matrix_c
```

```python
# adding nd arrays of different dimensions involves the intermediate step of broadcasting i.e. converting array_b in the same shape as array_c by
# 'blowing' out array_b and adding to array_c element wise
np.add(array_b, matrix_c)
```

### Type Casting

```python
# type casting involves taking every element of an array and changing it to a specified datatype
# type casting changes the inputs before performing the element wise operations
np.add(array_b, matrix_c, dtype = np.float64)
```

### Running over an Axis

```python
np.mean(matrix_c, axis = 0) # perform the mean operation over each column in the matrix
```
