```python
import numpy as np

array = np.array([-3, -57, 12, 43, 74, 75, 0, -6, 8])
matrix = np.array([[2, 6], [-5, 0], [11, -4]])

print(matrix[2,0]) # Print the element on row 3 and column 1
print(array[-2]) # Print the penultimate (second to last) element
print(matrix[:,1]) # Print the entire second column
```
