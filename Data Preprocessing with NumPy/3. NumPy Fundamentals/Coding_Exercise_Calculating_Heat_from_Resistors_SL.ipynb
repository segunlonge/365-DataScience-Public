```python
import numpy as np

# Currents through the resistors (in amperes)
currents = np.array([2, 0.5, 3, 2.6])

# Resistances of the resistors (in ohms)
resistances = np.array([3, 8, 1.3, 4.5])

# Calculate the power dissipated in each resistor using element-wise operations
# P = I^2 * R
currents_squared = currents * currents
power_dissipated = currents_squared * resistances

# Print the resulting array of power dissipated
print(power_dissipated)
```
