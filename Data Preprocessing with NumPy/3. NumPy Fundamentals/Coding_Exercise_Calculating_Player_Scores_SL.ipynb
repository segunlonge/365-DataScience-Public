```python
import numpy as np

# Define score values for Particles, Shards, and Orbs
score_weights = np.array([2, 5, 11]) # Fill in


# Number of gathered collectibles for each player: [# of Particles, # of Shards, # of Orbs]
player_collectibles = np.array([[14, 7, 3], [21, 15, 8], [13, 10, 1], [16, 14, 6], [13, 5, 10], [28, 8, 3], [17, 13, 9], [20, 11, 5]])

# Use NumPy's elementwise properties to calculate the scores by type of collectible
player_scores_by_type = score_weights*player_collectibles

print(player_scores_by_type)

# Find the total player scores (you can use np.sum())
player_scores = np.sum(player_scores_by_type, axis=1) # across rows
print(player_scores)
```
