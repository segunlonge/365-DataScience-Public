```python
import numpy as np

lending_co_data_numeric_NAN = np.genfromtxt("Lending-company-Numeric-NAN.csv", delimiter = ';')
lending_co_data_numeric_NAN
```

```python
# coordinate location of nana values
np.argwhere(np.isnan(lending_co_data_numeric_NAN))
```

```python
# Find the row indices containing NaNs
nan_rows = np.unique(np.argwhere(np.isnan(lending_co_data_numeric_NAN))[:, 0]) # The slicer takes on the row indexes
```

```python
nan_rows
```

```python
# Delete all those rows at once and reassign to the same variable
lending_co_data_numeric_NAN = np.delete(lending_co_data_numeric_NAN, nan_rows, axis=0)
```

```python
np.set_printoptions(suppress = True) # This option suppresses scientific notation to get a more readable output
lending_co_data_numeric_NAN
```

```python
#sort the n-d array in place using the sort method rather than the sort function
lending_co_data_numeric_NAN[:,5].sort()
lending_co_data_numeric_NAN
```
