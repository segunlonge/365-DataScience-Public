```python
# Find the mean of the unaltered dataset
data_nan = np.genfromtxt("Lending-company-Numeric-NAN.csv", delimiter = ";", skip_header = 0)
data_mean = np.nanmean(data_nan, axis = 0).round(2)
#data_nan
data_mean
```

```python
# Import data and fill all NAN values with -1
data_filled = np.genfromtxt("Lending-company-Numeric-NAN.csv", 
                            delimiter = ";", 
                            skip_header = 0,
                            filling_values = 0)

data_filled
```

```python
# Replace 0 with the mean in column 1
data_filled[:, 0] = np.where(data_filled[:, 0] == 0, data_mean[0], data_filled[:, 0])
data_filled
```
