## Exception handling

In this exercise, you will work with a given Python script that generates an error. Your task is to inspect the error and add a try block to handle the exception. Instead of fixing the error itself, use a try-except structure to catch the exception and print its message.

Important: Do not modify any other print statements in the code, as they are essential for validating the output.

### Skeleton code

```python
l = [3, 0, 5, 2, 2, 1, 9, 12, 6, 4, 2, 13, 10, 8, 3, 3]

sum = 0
print("Entering loop")

for i in range(len(l)+1):
  print(i)
  sum = sum + l[i]

print("The sum is: " + str(sum))
```

### Solution code

```python
l = [3, 0, 5, 2, 2, 1, 9, 12, 6, 4, 2, 13, 10, 8, 3, 3]

sum = 0
print("Entering loop")

try:
    for i in range(len(l)+1):
      print(i)
      sum = sum + l[i]
except IndexError as err:
    print(err)
    print("The sum is: " + str(sum))
```
