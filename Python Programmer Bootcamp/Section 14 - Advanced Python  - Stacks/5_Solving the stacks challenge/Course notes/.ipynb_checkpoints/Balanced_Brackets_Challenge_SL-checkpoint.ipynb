# Balanced Brackets Challenge

Create a function that will take an input that's a string which is made of up standard brackets i.e. {[( which are in a particular order. The function then needs to check whether they are in the right order (that is balanced) e.g.:

**{[()]}** is balanced but **{([)]}** is not balanced

Hints: 
1. The solution will need to use a stack
2. How do you fill the stack with the brackets and at what point do you check when they are balanced
3. Write down a pseudo solution before coding

### Idea 2

```python
def Balanced_Brackets_Function(string_input):
    stack = []
    brackets = {')': '(', ']': '[', '}': '{'} # key-value pair of brackets and their opposites

    for bracket in string_input: # search each character (opening bracket) in the input string
        if bracket in '([{':
            stack.append(bracket) # if found append
        elif bracket in ')]}': # if above if statement returns false (then search for the closing brackets)
            if not stack or stack[-1] != brackets[bracket]: # then if stack is not empty or last item in stack does not equal current bracket selection
                                                            # e.g. current stack="([" current bracket="]" and it's value "["
                return False # will return false with unmatched brackets
            stack.pop() # remove current bracket matches bracket at end of stack remove

    return len(stack) == 0 # by the end of the routine if brackets are matching they should all have been popped from the stack leaving it empty
                            # and hence true

print(Balanced_Brackets_Function("([]){}"))
            
            
        
```
