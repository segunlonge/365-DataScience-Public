## Actual code

```python
def linear_search_index(my_list, item):
  i = 0
  
  while i < len(my_list):
    if my_list[i] == item:
      return True, i
    i+=1
  
  return False, None


print(linear_search_index([5, 8, 12, 7, 6, 18], 12))
print(linear_search_index([5, 8, 12, 7, 6, 18], 3))
```
