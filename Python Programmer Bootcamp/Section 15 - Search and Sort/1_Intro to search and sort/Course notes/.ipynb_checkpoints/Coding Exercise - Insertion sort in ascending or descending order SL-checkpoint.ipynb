## Completed code

```python
def insertion_sort(my_list, reverse=False):
  
  # TODO: Write an insertion sort that allows for reverse order sorting
    n = len(my_list)
    for i in range(1,n): # start from the 2nd (current) index not the first
        value = my_list[i] # get the value of the 2nd index
        j = i # swap the index as j would be used for tracking the item to the left of the current index i 
        while j > 0 and my_list[j-1] > value: # check if the item to the left of the current index is greater than the item in the current index i
            my_list[j] = my_list[j-1] # if so move it to the right (as it is bigger)
            j = j - 1 # set to the index value left
        my_list[j] = value # move the current index value to the left (as it is smaller)

    if reverse == True:
        my_list.reverse() # if the parameter is True return the sorted list as reversed
        return my_list
    else:
        return my_list


# Test the function
print(insertion_sort([3, 1, 4, 1, 5]))
print(insertion_sort([10, 2, 8, 6, 7], reverse = True))
```
