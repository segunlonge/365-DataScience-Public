```python
def binary_search_recursive(my_list, item):
    if len(my_list) == 0:
        return False
    else:
        midpoint = len(my_list) // 2
        # If item found at midpoint return True
        if my_list[midpoint] == item:
            return True
        elif my_list[midpoint] < item:
            # Chop list to the right of midpoint
            return binary_search_recursive(my_list[midpoint+1:], item)
        else:
            # Chop list to the left of midpoint
            return binary_search_recursive(my_list[:midpoint], item)

# Testing the function
print(binary_search_recursive([1, 3, 5, 7, 9], 7))  # True
print(binary_search_recursive([1, 3, 5, 7, 9], 6))  # False
```
