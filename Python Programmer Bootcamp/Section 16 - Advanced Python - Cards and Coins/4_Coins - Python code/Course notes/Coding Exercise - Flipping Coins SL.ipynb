## Coding Exercise: Flipping coins

Create a program to simulate the coin flipping algorithm outlined in the previous video.<br>

Print a list containing the positions (not indices) of the coins that remain heads up when simulated with 1000 coins.<br>
(The output should be of the form [1, 4, 9])<br>


A reminder about the workings of the algorithm:<br>
    - Start with N coins heads up.<br>
    - Flip the 2nd, 4th, 6th... coin.<br>
    - Flip the 3rd, 6th, 9th... coin.<br>
    - Flip the 4th, 8th, 12th... coin.<br>
      ...<br>
    - Flip the nth coin.<br>
    - Take note of the final configuration of the coins.<br>

## Lesson Solution

```python
def flip_coins(n):
  # Function that simulates the coin flipping algorithm with an array of coins, where:
  # 1 - heads, 0 - tails
  
  # Generate array of n heads coins
  coins = [1]*n 
  
  # The first loop controls the step size (i)
  for i in range(2, n+1):
    for j in range(i-1,n,i):    # Iterate through the correct coins
      coins[j] = 1 - coins[j]   # Flip the coin at that position
  
  return coins


def find_positions(l, value = 1): 
  # Function to find the positions (not indices) of elements in a list,
  # that correspond to a particular value (by default 1, for heads)
  
  pos = list()
  for i in range(len(l)):
    if l[i] == value:
      pos.append(i+1)
  
  return pos


# Print the positions of heads for 1000 coins
print(find_positions(flip_coins(10)))
```
