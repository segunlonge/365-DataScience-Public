        - add all of the digits from those products
    2. Add the sum of the remaining digits (the digits not multiplied by 2)

## Solution code

```python
def card_number_val(number):
  
  digits = [int(char) for char in str(number)] # Break appart the number into digits
  
  doubled_digits = [2*d for d in digits[-2::-2]] # obtain the digits from step 1 and double them
  single_digits = digits[-1::-2] # obtain the rest of the digits (from step 2)
  
  # Find the different sums of digits
  sum_doubled_digits = sum([sum([int(char) for char in str(item)]) for item in doubled_digits])
  sum_single_digits = sum(single_digits)
  sum_total = sum_doubled_digits + sum_single_digits
  
  # Check the final sum for validity
  if sum_total % 10 == 0:
    return True
  else:
    return False

# Call the function with the following credit card numbers
print(card_number_val(5610591081018250))
print(card_number_val(4222222222222))
print(card_number_val(30569309026904))
```
