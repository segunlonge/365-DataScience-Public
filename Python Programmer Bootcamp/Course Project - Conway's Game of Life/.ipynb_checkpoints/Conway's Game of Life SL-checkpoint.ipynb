## Solution code

```python
class GameOfLife(object):  

# First task
    # Initialize a 2D list with dimensions x_dim by y_dim filled with zeros.
    '''
    Create a class constructor which will construct and initialise the variables in the parenthesis for each object created by the GameOfLife class
    Generating the 2d list with list comprehension (a way to create a new list from an iterable)
    '''
    def __init__(self, x_dim, y_dim):
        
        '''
        Parameters:
        x_dim: An integer representing the lenght of the x dimension of the grid
        y_dim: An integer representing the lenght of the y dimension of the grid
        '''

        self.x_dim = x_dim
        self.y_dim = y_dim
        self.life_grid = [[0 for _ in range(y_dim)] for _ in range(x_dim)]

# Second task
    # Implement a getter method for your grid.
    # print(self.life_grid) # prints the grid but does not return the grid as an object
    '''
    A def (a function) inside a class such as GameOfLife is simply a method (a function within a class)
    '''
    def get_grid(self):
        
        '''
        Returns the current life_grid object

        Returns:
        The life grid that was constructed from the class constructor (the __innit__ method)
        The grid is initially filled with 0s i.e. all dead cells
        '''
        return self.life_grid

# Third task
    # Implement a method to print out your grid in a human-readable format.
    '''
    Construct a print_grid() method to print the grid in a 'pretty' human readable format:
    '''
    def print_grid(self):
        '''
        Prints the current life_grid object in a readable format
        '''

        self.get_grid()
        # print(len(test_grid.get_grid()))
        # print(len(test_grid.get_grid()[0]))

        rows = len(self.get_grid()) # Count the number of rows
        columns = len(self.get_grid()[0]) # Count the number of elements only in the first row of the grid

        for r in range(rows):
            '''
            Return the full 2D grid and select the rth row then convert each cell in the row to a string using the generator expression
            {str(cell) for cell in self.get_grid()[r]}. Also join the character "|" onto each cell value.
            '''
            print("|".join(str(cell) for cell in self.get_grid()[r])) # Print "|" after each cell value in the life_grid
            if r < rows - 1: # print the character "-" except after the last row
                print("-" * (columns * 2 - 1)) # columns times 2 means 1 "-" character for the cell value and 1 for the separator minus 1 for the ends
                
# Fourth task
    # Given a list of 2D coordinates (represented as tuples/lists with 2 elements each),
    # set the corresponding elements in your grid to 1 (alive)
    '''
    Implement the populate_grid() method, which allows you to initialize the game state by specifying which cells in the grid should start as 'alive'
    (represented by a 1) which will help set stage for the Game of Life to evolve.
    '''
    def populate_grid(self, coord):
        '''
        Populates the game grid with live cells at the inputted coordinates
        
        Parameters:
        coord: A list of tuples with each representing the (x, y) coordinates of a live cell.
        
        Returns:
        The updated life_grid with the new live cells i.e. 1
        '''

        self.coord = coord
        rows = len(self.get_grid())
        cols = len(self.get_grid()[0])
        '''
        Loop through the inputted coordinates for x and y and check that none of the coordinates are outside the scope of the grid
        then populate the value of that location with 1
        '''
        for x, y in coord:
            if 0 <= x < rows and 0 <= y < cols:
                self.get_grid()[x][y] = 1

# Fith task
     # Implement the logic to update the game state according to the rules of Conway's Game of Life.
    '''
    Create a method which iterates over each cell in the grid, evaluates it's neighbours, and updates the cell's state based on the 
    Game of Life rules (see above)
    '''
    def make_step(self):
        '''
        Updates the Game of Life using the specificed rules in one step.
        
        Returns:
        The updated life_grid after one step.
        '''       
        
        # Initialising the sumgrid variable
        self.new_grid = [[0 for _ in range(self.y_dim)] for _ in range(self.x_dim)]

        # Get length of rows and column
        rows = len(self.get_grid()) # Count the number of rows
        columns = len(self.get_grid()[0]) # Count the number of elements only in the first row of the grid
        
        # Loop through each cell
        for i in range(rows):
            for j in range(columns):
        
                neighbours = []
                sum_neighbours = 0
        
                # Loop through the neighbours
                for di in (-1, 0, 1): # offsets i.e. for i one row up (-1), same row (0) and one row down (1)
                    for dj in (-1, 0, 1): # offsets i.e. for j one column left (-1), same column (0) and one column down (1)
        
                        # Skip the cell itself
                        if di == 0 and dj == 0:
                            continue
        
                        # adjust the current cell using the offsets
                        ni, nj = i + di, j + dj
        
                        # Check bounds to make sure loop doesn't go outside boundary
                        if 0 <= ni < rows and 0 <= nj < columns:
                            neighbours.append(self.get_grid()[ni][nj]) # After loop neighbours now contains the values of the 8 surrounding cells
        
                
                # print(f"Cell ({i},{j}) neighbours:", neighbours, sum(neighbours))
                # print(f"Cell value {self.get_grid()[i][j]} and sum:", sum(neighbours))
                sum_neighbours == sum(neighbours)

                # The rules of the Game of Life which determines which cells live or die
                if self.get_grid()[i][j] == 1:
                    if sum(neighbours) < 2:
                        #print(f"Cell {self.get_grid()[i][j]} dies")
                        self.new_grid[i][j] = 0
                    if sum(neighbours) == 2 or sum(neighbours) == 3:
                        #print(f"Cell {self.get_grid()[i][j]} lives")
                        self.new_grid[i][j] = 1
                    if sum(neighbours) > 3:
                        #print(f"Cell {self.get_grid()[i][j]} dies")
                        self.new_grid[i][j] = 0
                elif self.get_grid()[i][j] == 0: 
                    if sum(neighbours) == 3:
                        #print(f"Cell {self.get_grid()[i][j]} lives")
                        self.new_grid[i][j] = 1
                        
        # print("State of grid before the step is applied")                
        # print(self.get_grid()) #state of grid before the step is applied

        # This assignment does not create a new grid but instead creates another name that refers to lifegrid 2D list
        grid = self.get_grid() # Call the method get_grid() to get the life grid object, and store the returned grid in a local variable called grid

        # Update the local variable grid in accordance with the game of life rules
        for i in range(rows):
            for j in range(columns):
                grid[i][j] = self.new_grid[i][j]

# Sixth task
    # Implement a method that applies the make_step method n times.
    '''
    Create a function to repeatedly apply the rules of Game of Life i.e. call the make n step method so as to evolve the game
    '''
    def make_n_steps(self, n):
        '''
        AApplies the rules of the Game of Life by n steps.

        Parameters:
        n: The number of times to evolve the Game of Life

        Returns:
        The updated life_grid after n steps.
        '''        

        self.n = n

        for _ in range(n): # for _ is virtually the same as e.g. for i
            
            if _ == 0:
                #print("State of grid before the step is applied")            
                #print(self.get_grid()) #state of grid before the step is applied
                pass

            self.make_step()

        print(f"The game iterated through {n} steps")

# Seventh task
    # Draw the current state of the grid.
    '''
    Create a plot of the grid using matplotlib and colour code alive and dead cells
    '''
    def draw_grid(self):
        
        import matplotlib.pyplot as plt

        # Get length of rows and column
        rows = len(self.get_grid()) # Count the number of rows
        columns = len(self.get_grid()[0]) # Count the number of elements only in the first row of the grid

        # Initialize two empty lists for the x and y coordinates
        x = []
        y = []
        
        # Append the all the coordinates of the grid into x and y lists
        for i in range(rows):
            for j in range(columns):
                x.append(j)
                y.append(i)

        # Flatten the 2D life grid into a 1D list (needed for colour coding the grid)
        flat_life_grid = [value for row in self.get_grid() for value in row]
        # print(flat_life_grid)

        # plot
        plt.figure()
        colors = (flat_life_grid)
        plt.scatter(x, y, c=colors, cmap='coolwarm')
        scatter = plt.scatter(x, y, c=colors, cmap='coolwarm')
        #plt.legend(*scatter.legend_elements())
        handles, labels = scatter.legend_elements()
        labels = [f"Cell State = {lab}" for lab in labels]
        # plt.legend(handles, labels)
        # place the legend outside the boxplot
        plt.legend(
            handles,
            labels,
            loc="center left",
            bbox_to_anchor=(1.0, 0.5)
        )
        # plt.xlabel("x")
        # plt.ylabel("y")
        plt.title("Current state of Game of Life\n (Cell State 0 = Dead / Cell State 1 = Alive)")
        plt.show()
            
        



                
```

```python
class GameOfLife(object):  
    '''
    The GameOfLife class implements the famous "Conway's Game of Life".
    
    The game is a zero-player game, meaning that its evolution is determined by its initial state,
    requiring no further input. One interacts with the Game of Life by creating an initial 
    configuration and observing how it evolves.
    '''
    
    def __init__(self, x_dim, y_dim):
        '''
        Initializes a new instance of the GameOfLife class with a grid of given dimensions.
        
        Parameters:
        x_dim: The number of rows in the grid.
        y_dim: The number of columns in the grid.
        '''
        
        # Initialize life_grid as a 2D list filled with zeros. 
        # The dimensions of the grid are x_dim rows by y_dim columns.
        self.life_grid = [[0 for _ in range(y_dim)] for _ in range(x_dim)]
    
    def get_grid(self):
        '''
        Gets the current state of the life grid.

        Returns:
        The current state of the life grid.
        '''
        
        # Return the grid.
        return self.life_grid
    
    def print_grid(self):
        '''
        Prints the current state of the life grid in a readable format.
        '''
        
        # Iterate through each row in the grid
        for i in range(len(self.life_grid)):
            # For each cell in the current row, print its value followed by a pipe character
            # 'end = ' ' ensures that each cell value in a row is printed on the same line
            for j in range(len(self.life_grid[i])):
                print(self.life_grid[i][j], '|', end = ' ')
            # After printing all cell values in a row, print a newline character
            print()
            # Print a dashed line to visually separate the rows
            print('- '*6)

    def populate_grid(self, coord):
        '''
        Populates the game grid with live cells at the specified coordinates.
        
        Parameters:
        coord: A list of tuples. Each tuple represents the (x, y) coordinates of a live cell.
        
        Returns:
        The updated life_grid with the new live cells.
        '''
        
        # Loop through each set of coordinates
        for i in coord: 
            # Set the value at the specific coordinates (i[0], i[1]) in the life_grid to 1. 
            # This represents that the cell at these coordinates is alive.
            self.life_grid[i[0]][i[1]] = 1
        
        # Return the updated life_grid with the designated cells set to 1, representing live cells.
        return self.life_grid
            
    def make_step(self):
        '''
        Advances the game by one step according to the rules of the Game of Life.
        
        Returns:
        The updated life_grid after one step.
        '''
        
        # Get the number of rows in the life_grid list
        rows = len(self.life_grid)
        
        # Get the number of columns in the life_grid list
        columns = len(self.life_grid[0])
        
        # Initialize a variable to temporarily store the sum of a cell's neighboring cells
        temp_sum = 0.0
        
        # Create a new grid with the same dimensions as life_grid to store the sum of neighbors for each cell
        sum_grid = [[0 for i in range(columns)] for _ in range(rows)]
        
        # Iterate over each cell in the grid
        for i in range(rows):
            for j in range(columns):
                
                # Iterate over each of the cell's neighbors (including diagonally adjacent cells)
                for a in [i-1, i, i+1]:
                    for b in [j-1, j, j+1]:
                        # Do not include the cell itself in the sum of its neighbors
                        if a == i and b == j:
                            pass
                        else:
                            # If the neighbor is within the grid boundaries, add its value to the temporary sum
                            if 0 <= a < rows and 0 <= b < columns:
                                temp_sum += self.life_grid[a][b]
                            else:
                                pass
                            
                # Store the sum of a cell's neighbors in the corresponding position in sum_grid
                sum_grid[i][j] = temp_sum
                
                # Reset the temporary sum to 0 for the next cell
                temp_sum = 0.0
        
        # Iterate over each cell in the grid once more, this time to update the state of each cell
        for i in range(rows):
            for j in range(columns):
                # If the current cell is alive and has less than 2 or more than 3 live neighbors
                if self.life_grid[i][j] == 1:
                    if sum_grid[i][j] <= 1 or sum_grid[i][j] >= 4:
                        # The cell dies
                        self.life_grid[i][j] = 0
                    else:
                        pass
                 # If the current cell is dead and has exactly 3 live neighbors
                else:
                    if sum_grid[i][j] == 3:
                        # The cell becomes alive
                        self.life_grid[i][j] = 1
                    else:
                        pass
        
        # Return the updated grid
        return self.life_grid
    
    def make_n_steps(self, n):
        '''
        Advances the game by a specified number of steps.

        Parameters:
        n: The number of steps to advance the game by.

        Returns:
        The updated life_grid after n steps.
        '''
        
        # Iterate 'n' times, invoking the make_step function at each iteration to evolve the grid
        for i in range(n):
            self.make_step()
            
        # Return the updated grid after 'n' steps have been performed
        return self.life_grid
    
    def draw_grid(self):
        '''
        Visualizes the current state of the game grid using a scatter plot.

        This method generates x and y coordinates for each cell in the grid and 
        creates a scatter plot using matplotlib. Each cell is represented as a point
        in the plot. The color of the point indicates the state of the cell
        (alive or dead). The grid is displayed with inverted y-axis to match the 
        console output of the game.

        Returns:
        None
        '''
        
        # Get the number of rows and columns in the life_grid
        rows = len(self.life_grid) 
        columns = len(self.life_grid[0])
        
        # Initialize empty lists to hold the x and y coordinates for the plot
        x = []
        y = []
        
        # Populate the coordinate lists:
        # For each cell in the grid, append the row number to x and column number to y
        for i in range(columns):
            for j in range(rows):
                x.append(j)
                y.append(i)
        
        # Create a new figure with specified size
        fig, ax = plt.subplots(figsize=(9, 9))
        
        # Plot the state of each cell as a colored square marker on the scatter plot
        # Alive cells (value=1) and dead cells (value=0) will be colored differently
        plt.scatter(x, y, s = 150, c = self.life_grid, edgecolors = '#000000', marker = 's')
        
        # Set the axis limits to properly contain the grid
        plt.xlim(-1, rows)
        plt.ylim(-1, columns)
        
        # Invert the y-axis so that the plot visually matches the printed version of the grid
        ax.invert_yaxis()
        
        # Display the plot
        plt.show()
```
