# Coding exercise - Replace email domains

You are provided with a list of emails and a string new_domain. Your task is to use regular expressions to replace the domain of each email in the list with new_domain and print the updated list of emails.

For this exercise, you may find it helpful to check the regular expressions documentation:<br>
https://docs.python.org/3/library/re.html<br> 
Specifically, the re.sub() function may be particularly helpful:<br>
https://docs.python.org/3/library/re.html#re.sub

## Skeleton code

```python
import re

emails = ["john.doe@example.com", "jane.smith@test.org", "contact@company.net"]
new_domain = "365datascience.com"
```

## Completed code

```python
import re

'''
Pattern Explanation:
@: match the literal @ symbol
(.+): group(1) captures everything after @
Note (group(0) captures everything including @
'''

pattern = re.compile(r'@(.+)')
new_domain = "365datascience.com"

emails = ["john.doe@example.com", 
          "jane.smith@test.org", 
          "contact@company.net"]

#domains = [pattern.search(e).group(1) for e in emails if pattern.search(e)]
#print(domains)

for e in emails:
    print(pattern.search(e).group(1))

print('')

# Replace everything after '@' with the new domain
# re.sub(pattern, repl, string, count=0, flags=0)
updated_emails = [re.sub(pattern, '@' + new_domain, e) for e in emails]

print("Updated emails:")
for e in updated_emails:
    print(e)
```

## Course solution

```python
import re

emails = ["john.doe@example.com", "jane.smith@test.org", "contact@company.net"]
new_domain = "365datascience.com"

pattern = r'@\w+\.\w+'
new_emails = [re.sub(pattern, f'@{new_domain}', email) for email in emails]
print(new_emails)
```
