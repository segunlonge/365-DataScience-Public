# Coding Exercises - Validate Python variable names

In Python, a valid variable name must follow these rules:<br>
    - It can only contain letters(a-z; A-Z), numbers (0-9), and underscores (_).<br>
    - It cannot contain whitespace.<br>
    - It cannot start with a number.<br>
    - It cannot be a Python keyword (e.g., if, else, for, etc.).<br>

In this exercise, you'll write a function **variable_name_validator(name)** that uses regular expressions to check if a given string name can be a valid Python variable **name**. The output of this function should be **True** if the variable name is valid and **False** otherwise.

To check if a string is a Python keyword, you can use the **keyword** module from the Python standard library. The **keyword.iskeyword(s)** method will return **True** if the provided string is a Python keyword.

Finally, check your function with the provided test strings at the end.

*Hint: You can transform the result of re.match() to a boolean directly - e.g. bool(pattern.match(s)).*

```python
import re
import keyword

'''
REGEX pattern which checks that:
1. [A-Za-z_] = first character may contain upper case, lower case or underscore but cannot be a number
2. [A-Za-z0-9_]* = then the rest of the character can be alpha numeric and contain an underscore 
3. ^ (anchor) ensures the match begins at the very start of the string
4. $ Ensures the match ends at the very end of the string e.g. matching "valid_name123!" instead of "valid_name123". That is the match fails 
if there’s any extra character after the allowed ones
'''

pattern = re.compile(r'^[A-Za-z_][A-Za-z0-9_]*$')

def variable_name_validator(name):
  # TODO: Check if the provided string is a valid Python variable name
    return bool(pattern.fullmatch(name)) and not keyword.iskeyword(name)
    
# Test the function
print(variable_name_validator("test_var"))
print(variable_name_validator("2var"))
print(variable_name_validator("_underscore"))
print(variable_name_validator("class"))
print(variable_name_validator("myVar123"))
print(variable_name_validator("myVar123 "))
print(variable_name_validator("_"))
```

```python
import re

pattern = re.compile(r'^[A-Za-z_][A-Za-z0-9_]*$')

print(bool(pattern.match("valid_name123!")))  # True
```
