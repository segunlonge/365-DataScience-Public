# Coding Exercise - Formating strings

In this exercise, you will work with the mathematical constant $e$ (Euler's number), available through Python’s math module as math.e. Your task is to format and print various representations of $e$ by applying different formatting rules.

Each output should begin with the string "e is equal to ".

Follow this with each of the specified formatted versions of e:<br>
1. Display e with no modifications.<br>
2. Display e with a precision of 4 decimal places.<br>
3. Display e with exactly 4 significant digits.<br>
4. Display e with 2 decimal places and a leading + (plus) sign.<br>
5. Display e with 2 decimal places, right-aligned within a 13-character width.<br>
6. Display e with 2 decimal places, left-aligned within a 13-character width.<br>

```python
import math

# Original value of e
e = math.e

# 1. Display e with no modifications
print(f'e is equal to {e}')

# 2. Display e with a precision of 4 decimal places
print(f'e is equal to {e:.4f}')

# 3. Display e with exactly 4 significant digits
print(f'e is equal to {e:.4g}')

# 4. Display e with 2 decimal places and a leading '+' (plus) sign
print(f'e is equal to {e:+.2f}')

# 5. Display e with 2 decimal places, right-aligned within a 13-character width
print(f'e is equal to {e:13.2f}')

# 6. Display e with 2 decimal places, left-aligned within a 13-character width
print(f'e is equal to {e:<13.2f}')
```
