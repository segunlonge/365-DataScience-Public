```python
from math import ceil
message_decrypted = ''
key = 6
message = "Lnh egofa nurP nnyiits"
message_length = (len(message))
message_ceil = ceil(message_length / key)
# create an empty grid to hold the encryted message
message_grid = [["" for _ in range(key)] for _ in range(message_ceil)]
# num_empty_cells =  message_ceil * key - message_length
num_empty_cells = (message_ceil * key) - message_length

'''
An iterator is an object that allows you to traverse (iterate over) elements of a collection (like a list, tuple, or string) one at a time, 
without needing to know the underlying structure.
'''

message = list('Lnh egofa nurP nnyiits')
message_split = list(message)
#print(message_split)

iterator = iter(message_split)

# fill the created grid column by column 
for j in range(key):
    for i in range(message_ceil):
        # skips the two empty cells in the bottom right corner of the message_grid
        if j >= key - num_empty_cells and i == message_ceil - 1:
            continue # skip the next iterations if the shaded cells in the bottom right corner is reached rather than give an error
        message_grid[i][j] = next(iterator)

message_decrypted = "".join("".join(i) for i in message_grid)

print(message_decrypted)

# print list row by row into a grid
for i in message_grid:
    print(i)
```

Incorporate decryption method into Class

```python
from math import ceil

class TranspositionCipher(object): 
        
    def __init__(self, key, message):
        self.key = key
        self.message = message
        
    def encrypt_message(self, message):
        message_length = len(message)
        chars = list(message)
        encrypted_message = ""
        for j in range(self.key):
            for i in range(ceil(len(message)/self.key)):
                index = j + i * self.key
                if index < message_length:
                    encrypted_message += chars[index]
        return encrypted_message

    def decrypt_message(self, message):
        # Store each character in a list
        message_split = list(message)
        # Compute the total length of the message
        message_length = len(message)
        # Calculate the ceiling value of the ratio
        # of the message length to the key size
        message_ceil = ceil(message_length / self.key)
        # Calculate the number of empty cells in the
        # decryption grid by substracting the actual
        # message length from the total size of the grid
        # num_empty_cells = message_length - message_ceil * key
        num_empty_cells = (message_ceil * self.key) - message_length

        
        #Initialize a 2D grid with empty strings, which will be filled with the message 
        #characters for decryption
        message_grid = [["" for _ in range(self.key)] for _ in range(message_ceil)]

        iterator = iter(message_split)

        # fill the created grid column by column 
        for j in range(self.key):
            for i in range(message_ceil):
                # skips the two empty cells in the bottom right corner of the message_grid
                if j >= self.key - num_empty_cells and i == message_ceil - 1:
                    continue # skip the next iterations if the shaded cells in the bottom right corner is reached rather than give an error
                message_grid[i][j] = next(iterator)

        message_decrypted = "".join("".join(i) for i in message_grid)
        return message_decrypted
```
