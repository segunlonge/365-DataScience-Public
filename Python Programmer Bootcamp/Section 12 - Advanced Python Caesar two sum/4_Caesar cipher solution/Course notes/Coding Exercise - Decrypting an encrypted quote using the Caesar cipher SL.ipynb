Coding Exercise - Decrypting an encrypted quote using the Caesar cipher

Using the Caesar cipher from the solution as it enrypts differently to mine

```python
alphabet = 'abcdefghijklmnopqrstuvwxyz'

def shift_amount(i):
    return i%26

def encrypt(text,required_shift):
    out_string = ''
    text = text.lower()
    for char in text:
        if char not in alphabet:
            out_string = out_string + char
        else:    
            alpha_index = alphabet.find(char)
            out_string = out_string + alphabet[shift_amount(alpha_index+required_shift)]
    return out_string
    
encrypted_quote = "bj fwj ozxy fs fiafshji gwjji tk rtspjdx ts f rnstw uqfsjy tk f ajwd fajwflj xyfw. gzy bj hfs zsijwxyfsi ymj zsnajwxj. ymfy rfpjx zx xtrjymnsl ajwd xujhnfq."

decrypted_quote = encrypt(encrypted_quote, -5)

print(decrypted_quote)
```
