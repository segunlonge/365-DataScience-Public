Create a playing card class that when instatiated will display a playing card

```python
class class_card(object):
    #In Python, __init__ is a special method used in classes, not in regular functions. It’s called the constructor
    #The method is used to initialize the objects attributes
    def __init__(self, value, suit):
        self.value = value
        self.suit = suit

    #Return the value of the instatiated card    
    def get_value(self):
        return self.value

    #Return the suit of the instatiated card 
    def get_suit(self):
        return self.suit

    #Return the combined value and suit of the instatiated card
    def print_card(self):
        my_card = f'Card suit: {self.suit} and Card value: {self.value}'
        return my_card

card = class_card(9, "Diamonds")
card.print_card()
```
