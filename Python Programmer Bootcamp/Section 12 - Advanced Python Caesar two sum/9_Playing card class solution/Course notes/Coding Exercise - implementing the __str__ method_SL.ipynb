Create a class called Book that represents a book with the attributes title, author, and year_published. Implement the __str__ method to provide a custom string representation for instances of the Book class in the format

(Title) by (Author) (Published in Year)


Recall that the __str__ method is a dunder method (short for "double underscore") that allows you to specify how an instance of your class should be displayed as a string.

```python
# Implement the Book class
class Book:
    def __init__(self, title, author, year_published):
        # Define instance attributes here
        self.title = title
        self.author = author
        self.year_published = year_published

    def __str__(self):
        # Return the custom string representation
        return f"{self.title} by {self.author} (Published in {self.year_published})"

# Test the Book class
book = Book("Beartown", "Fredrik Backman", 2016)
print(book)
```
