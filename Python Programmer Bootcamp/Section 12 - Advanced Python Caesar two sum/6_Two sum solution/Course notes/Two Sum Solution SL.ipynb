Write a function which takes in several integer inputs into a list, a target number and works out 
which two numbers from that list add up to the target number, retrun their index numbers.

If no such pair of numbers exit then return -1

*(Brute force solution; see given solution for more elegant code but less comprehensive)*

```python
numbers = input("Please enter some integers separated by space: ")
list_of_numbers = [int(num) for num in numbers.split()]
target_number = int(input("Please enter a target number: "))
# print(list_of_numbers)

list_a = []
list_b = [] 

def two_sum(list_of_numbers, target_number):
    for i in range(len(list_of_numbers)):
        for j in range(i + 1, len(list_of_numbers)):  # avoid repeats and self-pairing
            sum_of_two = list_of_numbers[i] + list_of_numbers[j]
            if sum_of_two == target_number:
                #print(f"{list_of_numbers[i]} + {list_of_numbers[j]} = {sum_of_two}")
                list_a.append(f"{list_of_numbers[i]} + {list_of_numbers[j]} = {sum_of_two}")
                list_b.append(f"{i} & {j}")

    if not list_a: #check if list is empty
        return -1
    else:
        print(f"The list of numbers which add up to: {target_number} are {list_a}")
        print(f"The index numbers are {list_b} respectively")

two_sum(list_of_numbers, target_number)    
```
