```python
import pandas as pd

# Step 1: Load the dataset and create a copy
data = pd.read_csv('Lending-company.csv', index_col='LoanID')
lending_co_data = data.copy()

# Step 2: Inspect the StartDate column
print("StartDate column before conversion:")
print(lending_co_data['StartDate'].head())

# Step 3: Convert the StartDate column to datetime with dayfirst=True
lending_co_data['StartDate'] = pd.to_datetime(lending_co_data['StartDate'], dayfirst=True)
print("\nStartDate column after conversion:")
print(lending_co_data['StartDate'].head())

# Step 4: Extract and print year, month, and day of the first entry
first_entry = lending_co_data['StartDate'].iloc[0]
print("Year:", first_entry.year)
print("Month:", first_entry.month)
print("Day:", first_entry.day)
```
