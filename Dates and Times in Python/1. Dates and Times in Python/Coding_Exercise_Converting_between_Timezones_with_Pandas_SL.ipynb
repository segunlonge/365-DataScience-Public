```python
import pandas as pd

my_string_datetime = '2023-03-12 14:00:00'

# Step 1: Create a datetime object with "America/New_York" timezone
ny_datetime = pd.to_datetime(my_string_datetime).tz_localize('America/New_York')
print("New York Time:", ny_datetime)

# print(ny_datetime.tzinfo)

# Step 2: Convert the datetime to UTC
utc_datetime = ny_datetime.tz_convert('UTC')
print("UTC Time:", utc_datetime)

# Step 3: Convert UTC datetime to "Europe/London" timezone
london_datetime = utc_datetime.tz_convert('Europe/London')
print("London Time:", london_datetime)
```
