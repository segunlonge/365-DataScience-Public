```python
from datetime import datetime

# Step 1: Convert the string to a datetime object using strptime
string_datetime = '15:45:20 2023-05-12'
parsed_datetime = datetime.strptime(string_datetime, '%H:%M:%S %Y-%m-%d')
print("Parsed DateTime:", parsed_datetime)

# Step 2: Display only the year from parsed_datetime using strftime
print("Year:", parsed_datetime.year)

# Step 3: Create a datetime object for an event date and time
event_datetime = datetime(2025, 11, 11, 18, 15, 45)

# Step 4: Format event_datetime into a specific string
# strftime is a function used in Python to format date and time objects into strings based on a specified pattern.
formatted_event_time = event_datetime.strftime('The event is at %H hour(s), %M minute(s), and %S second(s).')
print(formatted_event_time)


```
