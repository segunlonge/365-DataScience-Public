```python
import pandas as pd
import datetime
import pytz
from matplotlib import pyplot as plt
data = pd.read_csv('Sales-products-tz-exam-set.csv', index_col = 'SaleID')
sales_products = data.copy()
sales_products
```

```python
# Show a summary of the dataframe
sales_products.describe()
```

```python
# Time difference between the first and second Outdoors Shop sale
first_product = pd.to_datetime(sales_products['DateOfSale'].iloc[0], format='%d/%m/%Y')
#print(first_product)
second_product = pd.to_datetime(sales_products['DateOfSale'].iloc[1], format='%d/%m/%Y')
#print(second_product)
delta = (first_product - second_product)
print("Days inbetween:", abs(delta.days))
```

```python
# create Unit timestamp value in another column by combining DateOfSale and TimeOfSale
sales_products['DateTimeStamp'] = pd.to_datetime(sales_products['DateOfSale'] +" "+ sales_products['TimeOfSale'], dayfirst=True)
sales_products
```

```python
# Reorder the columns with DateTimeStamp after TimeOfSale
column_names_reordered = ['RetailerCountry', 'RetailerType', 'Product', 'DateOfSale', 'TimeOfSale', 'DateTimeStamp', 'Timezone']
sales_products = sales_products[column_names_reordered]
sales_products.head()
```

```python
# convert each date-time to a Unix timestamp
from datetime import datetime
# 'apply' applies the lambda function to each value in the DateTimeStamp column to convert it into a Unix timestamp
sales_products['UnixTimeStamp'] = pd.to_datetime(sales_products['DateTimeStamp']).apply(lambda x: x.timestamp())
sales_products
```
