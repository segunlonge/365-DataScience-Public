```python
import pandas as pd

# Create and display the required DataFrame.

# By using a list of dictionaries
data = [{'Employee':'John Doe', 'Monthly Salary':5200},
          {'Employee':'Jane Smith', 'Monthly Salary':4800},
          {'Employee':'Michael Roberts', 'Monthly Salary':5000},
          {'Employee':'Emily Davis', 'Monthly Salary':5300},
          {'Employee':'Chris Green', 'Monthly Salary':4600},
          {'Employee':'Jessica Wilson', 'Monthly Salary':4950},
          {'Employee':'Daniel Kim', 'Monthly Salary':5100},
          {'Employee':'Sophia Martinez', 'Monthly Salary':4700}]
df = pd.DataFrame(data)
df
```

```python
import pandas as pd

# Create and display the required DataFrame.

# By using a dictionary of pandas Series (named names and monthly_salary)

names = pd.Series(['John Doe', 'Jane Smith', 'Michael Roberts', 'Emily Davis', 'Chris Green', 'Jessica Wilson', 'Daniel Kim', 'Sophia Martinez'])
monthly_salary = pd.Series([5200, 4800, 5000, 5300, 4600, 4950, 5100, 4700])
data = {'Employee':names, 'Monthly Salary':monthly_salary}
df = pd.DataFrame(data)
df
```
